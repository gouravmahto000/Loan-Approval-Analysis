```python
1. C=1.0

Think of it as:

How strict should the model be?

Small C → More strict → Simpler model
Large C → Less strict → Tries harder to fit data

Example:

LogisticRegression(C=0.1)

vs

LogisticRegression(C=10)

For beginners:

C=1.0

is fine.

2. max_iter=100

Think of it as:

How many times can the model try to learn?

Example:

max_iter=100

means:

"Try up to 100 times."

If the model says:

ConvergenceWarning

increase it:

max_iter=1000
3. random_state=42

Think of it as:

Use the same randomness every time.

Without it:

model = LogisticRegression()

results can vary slightly.

With:

model = LogisticRegression(random_state=42)

you get reproducible results.
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(
    x_train,
    y_train
)
```

```python
Think of Random Forest as:

🌳 One Decision Tree = One person's opinion
🌲🌲🌲🌲🌲 Random Forest = 100 people's opinions combined

1. n_estimators = 100
n_estimators=100

Means:

Build 100 decision trees.

Example:

RandomForestClassifier(n_estimators=200)

creates 200 trees.

More trees:

Usually better accuracy
More training time
2. max_depth = None
max_depth=None

Means:

Trees can grow as deep as needed.

Example:

max_depth=5

Tree can only have 5 levels.

Think:

Depth = How many questions the tree can ask

Too deep → Overfitting

3. criterion='gini'
criterion='gini'

Used to decide:

Which question should the tree ask next?

Common options:

'gini'
'entropy'

For beginners:

gini

is fine.

4. min_samples_split = 2
min_samples_split=2

A node must have at least 2 samples before it can split.

Example:

10 customers

can split.

1 customer

cannot split.

5. min_samples_leaf = 1
min_samples_leaf=1

Minimum records required in a leaf node.

Example:

Approved

can contain even 1 customer.

6. max_features='sqrt'
max_features='sqrt'

Random Forest does NOT look at all columns at every split.

It randomly selects some columns.

Example:

If you have:

Age
Income
Credit Score
Loan Amount
City

it might only look at:

Income
Credit Score

for a particular split.

This makes trees different from each other.

7. bootstrap=True
bootstrap=True

Each tree gets a random sample of the data.

Example:

Tree 1:

Customers 1,2,3,4,5

Tree 2:

Customers 2,5,7,8,9

This randomness is what makes Random Forest powerful.
```
